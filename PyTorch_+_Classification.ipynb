```python
import torch #torch → main PyTorch library
import torch.nn as nn # nn → neural network tools, blueprint for neural networks
import torch.optim as optim # optim → optimization algorithms
```

```python
X = torch.tensor([[1.0], [2.0], [3.0], [8.0], [9.0], [10.0]])
y = torch.tensor([[0.0], [0.0], [0.0], [1.0], [1.0], [1.0]])
#Tensor → like NumPy array but smarter
```

```python
class SimpleClassifier(nn.Module):
    def __init__(self):
        super().__init__() #Required for PyTorch internals
        self.linear = nn.Linear(1, 1) # y = wx + b

    def forward(self, x):
        return torch.sigmoid(self.linear(x)) # sigmoid converts output into probability (0–1)
```

```python
model = SimpleClassifier()
criterion = nn.BCELoss() # Binary Cross Entropy Loss
optimizer = optim.SGD(model.parameters(), lr=0.1) # lr=0.1 → learning rate (step size)
```

```python
for epoch in range(1000):
    y_pred = model(X)
    loss = criterion(y_pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
```

```python
test = torch.tensor([[4.0]])
prediction = model(test)
print(prediction)
```
